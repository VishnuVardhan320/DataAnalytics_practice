```python
#smart fitness center Descriptive Analysis
import statistics as sa
mn=["Aman","Riya","karan","sneha","ravi"]
age=[15,16,17,19,20]
h=[2,4,3,1.5,5]
calb=[500,1000,700,300,2000]
for i in range(len(mn)):
  print(mn[i],age[i],h[i])
aw=sum(h)/len(h)
print("average age",aw)
```

```python
mi=min(h)
print(mi)
ma=max(h)
print(ma)
med=sa.median(calb)
print(med)
std=sa.stdev(calb)
print(std)
ind=calb.index(1000)
print(ind)
```

```python
import matplotlib.pyplot as plt
plt.bar(mn,h)
plt.show()
plt.bar(mn,calb)
plt.show()
plt.pie(h,labels=mn)
plt.show()
plt.pie(calb,labels=mn)
plt.show()
plt.boxplot(calb)
plt.show()
```
