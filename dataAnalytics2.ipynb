```python
import pandas as pd
d=pd.read_excel("sales1.xlsx")
data=pd.DataFrame(d)
sales=data.to_csv("sales.csv")
print(data)
```

```python
print("DataFrame Info:")
data.info()
print("\nDataFrame Description (Numerical Columns):")
display(data.describe())
print("\nMissing Values:")
display(data.isnull().sum())
```

```python
print("Unique Regions:", data['region'].unique())
print("Unique Countries:", data['country'].unique())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(data['sales'], kde=True, ax=axes[0])
axes[0].set_title('Distribution of Sales')

sns.histplot(data['prices'], kde=True, ax=axes[1])
axes[1].set_title('Distribution of Prices')

plt.tight_layout()
plt.show()
```

```python
sales_by_country = data.groupby('country')['sales'].sum().reset_index()
prices_by_country = data.groupby('country')['prices'].mean().reset_index()

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
sns.barplot(x='country', y='sales', data=sales_by_country, ax=axes[0], palette='viridis')
axes[0].set_title('Total Sales by Country')
axes[0].set_ylabel('Total Sales')

sns.barplot(x='country', y='prices', data=prices_by_country, ax=axes[1], palette='magma')
axes[1].set_title('Average Prices by Country')
axes[1].set_ylabel('Average Prices')

plt.tight_layout()
plt.show()
```

### 5. Sales and Prices by Region

```python
sales_by_region = data.groupby('region')['sales'].sum().reset_index()
prices_by_region = data.groupby('region')['prices'].mean().reset_index()

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
sns.barplot(x='region', y='sales', data=sales_by_region, ax=axes[0], palette='viridis')
axes[0].set_title('Total Sales by Region')
axes[0].set_ylabel('Total Sales')

sns.barplot(x='region', y='prices', data=prices_by_region, ax=axes[1], palette='magma')
axes[1].set_title('Average Prices by Region')
axes[1].set_ylabel('Average Prices')

plt.tight_layout()
plt.show()
```

```python
data["Amount"]=data["sales"]*data["prices"]

print(data)
plt.bar(data["region"],data["Amount"])
plt.show()
```

```python
#diagnostic analysis project
#smart E-Commerece cart
cus_n=["Aman","riya","k","s","r"]
internet_speed=[2,10,1,15,3]
payment_failure=[3,0,4,0,2]
delivery_charges=[15,10,5,6,10]
purchase_complete=[5,10,15,8,7]

cart_data = pd.DataFrame({
    'customer_name': cus_n,
    'internet_speed': internet_speed,
    'payment_failure': payment_failure,
    'delivery_charges': delivery_charges,
    'purchase_complete': purchase_complete
})

print("E-Commerce Cart Data:")
display(cart_data)
print("\nDataFrame Info:")
cart_data.info()
```

### Diagnostic Analysis: Internet Speed vs. Purchase Completion

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x='internet_speed', y='purchase_complete', data=cart_data, hue='customer_name', s=100)
plt.title('Internet Speed vs. Purchase Completion')
plt.xlabel('Internet Speed (Mbps)')
plt.ylabel('Purchase Complete (Number of Purchases)')
plt.grid(True)
plt.show()
```

### Diagnostic Analysis: Payment Failure vs. Purchase Completion

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x='payment_failure', y='purchase_complete', data=cart_data, hue='customer_name', s=100)
plt.title('Payment Failure vs. Purchase Completion')
plt.xlabel('Payment Failure (Count)')
plt.ylabel('Purchase Complete (Number of Purchases)')
plt.grid(True)
plt.show()
```
